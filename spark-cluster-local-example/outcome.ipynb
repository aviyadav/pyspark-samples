```python
from pyspark.sql import SparkSession
import pyspark.sql.functions as F

spark = (
    SparkSession.builder
    .appName("Read ETL Results")
    # local mode is fine for notebook viewing
    .master("local[*]")
    .getOrCreate()
)

CURATED = "data/results/curated_taxi_trips"      # host path
KPIS = "data/results/daily_zone_kpis"            # host path

curated = spark.read.parquet(CURATED)
kpis = spark.read.parquet(KPIS)

print("Curated schema:")
curated.printSchema()

print("KPIs schema:")
kpis.printSchema()

# Show curated sample
curated.orderBy(F.desc("total_amount")).show(10, truncate=False)

# Show KPI sample
kpis.orderBy(F.desc("trip_count")).show(20, truncate=False)

# If you want a nicer table: convert a SMALL sample to pandas
kpis_pd = kpis.orderBy(F.desc("trip_count")).limit(50).toPandas()
kpis_pd.head()
```
